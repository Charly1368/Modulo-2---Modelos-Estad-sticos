Ejercicio 1: Intervalo de confianza para proporción
Datos:

n = 225

x = 100

Nivel de confianza = 95%

Solución:

Proporción muestral: p̂ = x/n = 100/225 ≈ 0.4444

Error estándar: SE = sqrt(p̂*(1-p̂)/n) ≈ sqrt(0.4444*0.5556/225) ≈ 0.0332

Z para 95% ≈ 1.96

Intervalo: p̂ ± ZSE ≈ 0.4444 ± 1.960.0332 ≈ (0.379, 0.510)

Ejercicio 2: Datos normales
Datos: [1.2, 2.1, 2.3, 1.9, 3.0, 1.5, 2.8, 2.4, 1.7, 2.6, 2.0, 1.8, 3.1, 2.2, 2.5]

a) σ conocida = 2, nivel 90%

n = 15

x̄ ≈ 2.1933

Z para 90% ≈ 1.645

SE = σ/√n = 2/√15 ≈ 0.5164

Intervalo: 2.1933 ± 1.645*0.5164 ≈ (1.344, 3.042)

b) σ desconocida, nivel 90%

s ≈ 0.5126 (de la muestra)

t para 14 grados de libertad, 90% ≈ 1.761

SE = s/√n ≈ 0.5126/√15 ≈ 0.1323

Intervalo: 2.1933 ± 1.761*0.1323 ≈ (1.960, 2.426)

c) Intervalo para σ², nivel 90%

χ² inferior (α/2=0.05, df=14) ≈ 23.685

χ² superior (1-α/2=0.95, df=14) ≈ 6.571

Intervalo: [(n-1)s²/χ²sup, (n-1)s²/χ²inf] ≈ [140.2628/23.685, 140.2628/6.571] ≈ [0.155, 0.560]

Ejercicio 7: Toxafen en ratas
Datos:

Grupo bajo: n1=20, x̄1=190, s1=54

Grupo control: n2=23, x̄2=210, s2=32

Nivel 90%

Solución:

Diferencia: 190-210 = -20

SE = sqrt(s1²/n1 + s2²/n2) = sqrt(54²/20 + 32²/23) ≈ 13.77

Grados de libertad ≈ 30 (aproximación conservadora)

t ≈ 1.697

Intervalo: -20 ± 1.697*13.77 ≈ (-43.37, 3.37)

El intervalo incluye el 0, lo que sugiere que no hay diferencia significativa en el nivel de 90%. El toxafen podría no estar reduciendo significativamente el peso.

Ahora el código Python para verificar estos cálculos:

```python
import numpy as np
from scipy import stats

# Ejercicio 1
n1, x1 = 225, 100
p_hat = x1/n1
se1 = np.sqrt(p_hat*(1-p_hat)/n1)
z = stats.norm.ppf(0.975)
ci1 = (p_hat - z*se1, p_hat + z*se1)
print(f"Ejercicio 1 IC: {ci1}")

# Ejercicio 2
data = [1.2, 2.1, 2.3, 1.9, 3.0, 1.5, 2.8, 2.4, 1.7, 2.6, 2.0, 1.8, 3.1, 2.2, 2.5]
n2 = len(data)
xbar2 = np.mean(data)
s2 = np.std(data, ddof=1)

# 2a
sigma = 2
se2a = sigma/np.sqrt(n2)
z90 = stats.norm.ppf(0.95)
ci2a = (xbar2 - z90*se2a, xbar2 + z90*se2a)
print(f"Ejercicio 2a IC: {ci2a}")

# 2b
se2b = s2/np.sqrt(n2)
t90 = stats.t.ppf(0.95, n2-1)
ci2b = (xbar2 - t90*se2b, xbar2 + t90*se2b)
print(f"Ejercicio 2b IC: {ci2b}")

# 2c
chi2_low = stats.chi2.ppf(0.05, n2-1)
chi2_high = stats.chi2.ppf(0.95, n2-1)
ci2c = ((n2-1)*s2**2/chi2_high, (n2-1)*s2**2/chi2_low)
print(f"Ejercicio 2c IC: {ci2c}")

# Ejercicio 3
n3, xbar3, s3 = 250, 369, 50
se3 = s3/np.sqrt(n3)
ci3 = (xbar3 - z*se3, xbar3 + z*se3)
print(f"Ejercicio 3 IC: {ci3}")

# Ejercicio 4
n4, xbar4, sigma4 = 480, 12658, 2000
se4 = sigma4/np.sqrt(n4)

# 4a
z90_4 = stats.norm.ppf(0.95)
ci4a = (xbar4 - z90_4*se4, xbar4 + z90_4*se4)
print(f"Ejercicio 4a IC: {ci4a}")

# 4b
ci4b = (xbar4 - z*se4, xbar4 + z*se4)
print(f"Ejercicio 4b IC: {ci4b}")

# 4c
z99 = stats.norm.ppf(0.995)
ci4c = (xbar4 - z99*se4, xbar4 + z99*se4)
print(f"Ejercicio 4c IC: {ci4c}")

# Ejercicio 5
defects = [0,1,1,2,1,0,2,3,2,1,0,2,0,0,2,3,0,4,3,1,1,1,0,2,0,2,0,3,1,0]
n5 = len(defects)
xbar5 = np.mean(defects)
s5 = np.std(defects, ddof=1)
se5 = s5/np.sqrt(n5)
t95 = stats.t.ppf(0.975, n5-1)
ci5 = (xbar5 - t95*se5, xbar5 + t95*se5)
print(f"Ejercicio 5a Media: {xbar5}")
print(f"Ejercicio 5b Desv Est: {s5}")
print(f"Ejercicio 5c IC: {ci5}")

# Ejercicio 6
batteries = [2200,2290,2390,2410,2480,2500,2580,2700,2030,2100,2190,1600,1740,1900,1930,2000,1510,1470,1770,1710]
n6 = len(batteries)
xbar6 = np.mean(batteries)
s6 = np.std(batteries, ddof=1)
se6 = s6/np.sqrt(n6)
t90_6 = stats.t.ppf(0.95, n6-1)
ci6 = (xbar6 - t90_6*se6, xbar6 + t90_6*se6)
print(f"Ejercicio 6 Media: {xbar6}, Desv Est: {s6}")
print(f"Ejercicio 6 IC: {ci6}")

# Ejercicio 7
n7a, xbar7a, s7a = 20, 190, 54
n7b, xbar7b, s7b = 23, 210, 32
diff = xbar7a - xbar7b
se7 = np.sqrt(s7a**2/n7a + s7b**2/n7b)
df7 = (s7a**2/n7a + s7b**2/n7b)**2 / ((s7a**2/n7a)**2/(n7a-1) + (s7b**2/n7b)**2/(n7b-1))
t90_7 = stats.t.ppf(0.95, df7)
ci7 = (diff - t90_7*se7, diff + t90_7*se7)
print(f"Ejercicio 7 IC: {ci7}")
```
